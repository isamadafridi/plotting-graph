# **Plotting List Data with the matplotlib Package**

# **Plotting a Line Graph**

```python
import matplotlib.pyplot as plt
```

# Exaxmple

```python

x_coords = [0, 1, 2, 3, 4]
y_coords = [0, 3, 1, 5, 2]
plt.plot(x_coords, y_coords, marker = 'o')

# Set the axis limits.
plt.xlim(xmin=-1, xmax=10)
plt.ylim(ymin=-1, ymax=7)

# Add a title.
plt.title('My Plot')

# Add labels to the axes.
plt.xlabel('x axis')
plt.ylabel('y axis')

#Line on graph like graph paper
plt.grid(True)

plt.show()
```

# **Plotting a Bar Chart**

 You can use the bar function in the matplotlib.pyplot module to create a bar chart.
A bar chart has a horizontal X axis, a vertical Y axis, and a series of bars that typically
originate from the X axis.

```python
lef_edges = [10, 11, 12,13,14]
heights = [100, 200, 300, 400, 500]
bar_width = 1

plt.bar(lef_edges,heights, bar_width, color=('k', 'g', 'b', 'm', 'c'))
plt.show()
```

# **Plotting a Pie Chart**
 A pie chart is a graph that shows a circle that is divided into slices. The circle represents
the whole, and the slices represent percentages of the whole. You use the pie function in
the matplotlib.pyplot module to create a pie chart.

```python
# Create a list of values
values = [20, 60, 80, 40,50]

# Create a list of labels for the slices.
slice_labels = ['1st Qtr', '2nd Qtr', '3rd Qtr', '4th Qtr','5th Qtr']

# Create a pie chart from the values.
plt.pie(values, labels = slice_labels, colors=('r', 'g', 'b', 'c', 'k'))
# Set the title of the pie chart.
plt.title('Pie Chart') # calling the title function of plt and passing a string 'Pie Chart'
plt.show()
```
